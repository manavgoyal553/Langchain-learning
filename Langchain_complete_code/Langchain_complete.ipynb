```python
# ============================================================
# COMPLETE RAG PIPELINE — End to End
# ============================================================

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# STEP 1: Load
loader = TextLoader('Speech.txt')
documents = loader.load()
print(f'Step 1 — Loaded: {len(documents)} document(s)')

# STEP 2: Split
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = splitter.split_documents(documents)
print(f'Step 2 — Split into: {len(chunks)} chunks')

# STEP 3: Embed
embeddings = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)
print(f'Step 3 — Embedding model ready')

# STEP 4: Store
vector_store = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)
print(f'Step 4 — Vectors stored in FAISS')

# STEP 5: Retrieve
retriever = vector_store.as_retriever(search_kwargs={'k': 3})
user_query = 'What is the main message of the speech?'
results = retriever.invoke(user_query)
print(f'Step 5 — Retrieved {len(results)} relevant chunks')

print('\n' + '='*50)
print('QUERY:', user_query)
print('='*50)
for i, doc in enumerate(results):
    print(f'\nChunk {i+1}:')
    print(doc.page_content)
    print('-'*40)

# NOTE: In Week 5 you will add Step 6:
# context = results → combine with prompt template → send to LLM → get final answer
# That is the retrieval chain which completes the full RAG application
```
